```python
import torch
import torch.nn as nn 
input = torch.randint(5, (1, 6, 6), dtype = torch.float32)
input
```

```python
# define convoluational layer
conv_layer = nn.Conv2d(
    in_channels = 1,
    out_channels = 1, 
    kernel_size = 3,
    bias = False
)
conv_layer.weight
```

```python
init_kernel_weight = torch.randint(
    high = 2,
    size = (conv_layer.weight.data.shape),
    dtype = torch.float32
)
init_kernel_weight
```

```python
conv_layer.weight.data = init_kernel_weight
conv_layer.weight
```

```python
print(conv_layer(input))
```

<h1>Bias</h1>

```python
convolution_layer = nn.Conv2d(
    in_channels = 1,
    out_channels = 1,
    kernel_size = 3,
)
print(convolution_layer.bias)
```

```python
convolution_layer.bias = nn.Parameter(
    torch.tensor([1], dtype = torch.float32)
)
convolution_layer.bias
```

```python
output = convolution_layer(input)
print(output)
```

<h1>Padding</h1>

```python
input = torch.randint(5, (1, 4, 4), dtype = torch.float32)
print(input)
```

```python
init_kernel_weight = torch.randint(
    high = 2,
    size = convolution_layer.weight.data.shape,
    dtype = torch.float32
)
init_kernel_weight
```

```python
convolution_layer = nn.Conv2d(
    in_channels = 1, 
    out_channels = 1,
    kernel_size = 3,
    dtype = torch.float32,
    padding = 'same'
)
```

```python
convolution_layer.weight.data = init_kernel_weight
print(convolution_layer.weight)
```

```python
output = convolution_layer(input)
print(output)
```

<h2>Padding with parameter</h2>

```python
convolution_layer = nn.Conv2d(
    in_channels = 1, 
    out_channels = 1,
    kernel_size = (3, 3),
    bias = False,
    dtype = torch.float32,
    padding = (1, 2)
)
convolution_layer.weight.data = init_kernel_weight
output = convolution_layer(input)
print(output)
```

<h1>Stride</h1>

```python
convolution_layer = nn.Conv2d(
    in_channels = 1,
    out_channels = 2,
    kernel_size = 3,
    bias = False,
    dtype = torch.float32,
    padding = (2, 2),
    stride = (2, 2)
)
output = convolution_layer(input)
print(output)
```

<h1>Pooling</h1>

```python
input = torch.randint(5, (1, 6, 6), dtype = torch.float32)
input
```

```python
max_pool_layer = nn.MaxPool2d(
    kernel_size = 2,
    stride = (1, 2) # Truot theo hang, sau do den cot
)
output = max_pool_layer(input)
print(output)
```

<h2>Max Pool 1D</h2>

```python
input = torch.randint(5, (1, 6, 6), dtype = torch.float32)
input
```

```python
max_pool_layer_1D = nn.MaxPool1d(
    kernel_size = 3,
    stride = 3
)
output = max_pool_layer_1D(input)
print(output)
```

<h2>Average Pooling</h2>

```python
input = torch.randint(5, (1, 6, 6), dtype = torch.float32)
input
```

```python
avg_pool_layer = nn.AvgPool2d(
    kernel_size = (3, 2),
    stride = (2, 2)
)
output = avg_pool_layer(input)
print(output)
```

<h1>Flatten</h1>

```python
input = torch.randint(5, (1, 3, 2), dtype = torch.float32)
input
```

```python
flatten_layer = nn.Flatten()
output = flatten_layer(input)
print(output)
```
